```python
from pathlib import Path
import os

HOME = Path.home()

PROJECT_DIR = HOME / "practice_seg"
BASE_DIR = PROJECT_DIR

NNUNET_RAW = BASE_DIR / "nnUNet_raw"
NNUNET_PREPROCESSED = BASE_DIR / "nnUNet_preprocessed"
NNUNET_RESULTS = BASE_DIR / "nnUNet_results"

DATASET_DIR = NNUNET_RAW / "Dataset101_MedDDPM"

IMAGES_TR = DATASET_DIR / "imagesTr"
LABELS_TR = DATASET_DIR / "labelsTr"
IMAGES_TS = DATASET_DIR / "imagesTs"

# ============================================================
# 全班共享快取（課程伺服器 /home/pipcache）
# 若講師已準備好共享資料集，自動 link 進專案資料夾；
# 學生不用重新複製資料或重複 preprocessing
# ============================================================
SHARED = Path("/home/pipcache")

def link_shared(name, local, prepared):
    shared = SHARED / name
    if local.is_symlink():
        print(f"{name}: 已連結共享資料集 ({local})")
    elif local.exists():
        print(f"{name}: 本地資料夾已存在，沿用本地 ({local})")
    elif shared.is_dir() and prepared(shared):
        local.symlink_to(shared)
        print(f"{name}: 使用全班共享 {shared}")
    else:
        print(f"{name}: 無全班共享資料集，將在本機建立")

link_shared(
    "nnUNet_raw",
    NNUNET_RAW,
    lambda s: (s / "Dataset101_MedDDPM" / "dataset.json").is_file(),
)

link_shared(
    "nnUNet_preprocessed",
    NNUNET_PREPROCESSED,
    lambda s: (s / "Dataset101_MedDDPM" / "nnUNetPlans.json").is_file(),
)

for path in [
    PROJECT_DIR,
    BASE_DIR,
    NNUNET_RAW,
    NNUNET_PREPROCESSED,
    NNUNET_RESULTS,
    DATASET_DIR,
    IMAGES_TR,
    LABELS_TR,
    IMAGES_TS,
]:
    path.mkdir(parents=True, exist_ok=True)

os.environ["nnUNet_raw"] = str(NNUNET_RAW)
os.environ["nnUNet_preprocessed"] = str(NNUNET_PREPROCESSED)
os.environ["nnUNet_results"] = str(NNUNET_RESULTS)

print("資料夾建立完成：")
print(DATASET_DIR)

print("\nimagesTr :", IMAGES_TR)
print("labelsTr :", LABELS_TR)
print("imagesTs :", IMAGES_TS)
```

```python
from pathlib import Path
import shutil

HOME = Path.home()

SRC_IMAGES = HOME / "Public/train/med-ddpm-1/image"
SRC_MASKS = HOME / "Public/train/med-ddpm-1/mask"

DATASET_DIR = HOME / "practice_seg/nnUNet_raw/Dataset101_MedDDPM"

IMAGES_TR = DATASET_DIR / "imagesTr"
LABELS_TR = DATASET_DIR / "labelsTr"
IMAGES_TS = DATASET_DIR / "imagesTs"

if len(list(IMAGES_TR.glob("*.nii.gz"))) > 0:
    print("imagesTr 已有資料（全班共享資料集或先前已建立），跳過複製。")
else:
    IMAGES_TR.mkdir(parents=True, exist_ok=True)
    LABELS_TR.mkdir(parents=True, exist_ok=True)
    IMAGES_TS.mkdir(parents=True, exist_ok=True)

    print("Source images:")
    print(SRC_IMAGES)

    print("\nSource masks:")
    print(SRC_MASKS)

    assert SRC_IMAGES.exists(), f"找不到影像資料夾: {SRC_IMAGES}"
    assert SRC_MASKS.exists(), f"找不到 mask 資料夾: {SRC_MASKS}"

    def get_nii_files(folder):
        files = list(folder.glob("*.nii.gz"))
        files += list(folder.glob("*.nii"))
        return sorted(files)

    image_files = get_nii_files(SRC_IMAGES)
    mask_files = get_nii_files(SRC_MASKS)

    print("\n影像數量:", len(image_files))
    print("Mask 數量:", len(mask_files))

    assert len(image_files) == len(mask_files), (
        f"影像數量 {len(image_files)} != "
        f"mask 數量 {len(mask_files)}"
    )

    def strip_nii_suffix(path):
        name = path.name
        if name.endswith(".nii.gz"):
            return name[:-7]
        elif name.endswith(".nii"):
            return name[:-4]
        return path.stem

    image_map = {
        strip_nii_suffix(p): p
        for p in image_files
    }

    mask_map = {
        strip_nii_suffix(p): p
        for p in mask_files
    }

    image_keys = set(image_map.keys())
    mask_keys = set(mask_map.keys())

    missing_masks = sorted(image_keys - mask_keys)
    missing_images = sorted(mask_keys - image_keys)

    assert not missing_masks, f"找不到 mask: {missing_masks[:10]}"
    assert not missing_images, f"找不到 image: {missing_images[:10]}"

    case_keys = sorted(image_keys)

    for idx, case_key in enumerate(case_keys):

        case_id = f"seg_{idx:04d}"

        src_img = image_map[case_key]
        src_mask = mask_map[case_key]

        dst_img = IMAGES_TR / f"{case_id}_0000.nii.gz"
        dst_mask = LABELS_TR / f"{case_id}.nii.gz"

        shutil.copy2(src_img, dst_img)
        shutil.copy2(src_mask, dst_mask)

    print("\n完成！")
    print("Cases:", len(case_keys))
    print("imagesTr:", len(list(IMAGES_TR.glob("*.nii.gz"))))
    print("labelsTr:", len(list(LABELS_TR.glob("*.nii.gz"))))
```

```python
from pathlib import Path
import nibabel as nib
import numpy as np

HOME = Path.home()

LABELS_TR = (
    HOME
    / "practice_seg/"
      "nnUNet_raw/Dataset101_MedDDPM/labelsTr"
)

label_files = sorted(LABELS_TR.glob("*.nii.gz"))

assert len(label_files) > 0, "labelsTr 沒有資料"

allowed = {0, 1, 2}
needs_convert = []

for label_file in label_files:

    img = nib.load(label_file)

    data = np.asarray(img.dataobj)

    unique_before = np.unique(data)

    unexpected = set(unique_before.tolist()) - allowed

    assert not unexpected, (
        f"{label_file.name} 發現未知 label: {unexpected}"
    )

    if (data == 2).any():
        needs_convert.append(label_file)

if not needs_convert:
    print("所有 labelsTr 已是 0/1（全班共享資料集或已轉換過），跳過 2 -> 1。")
else:
    for label_file in needs_convert:

        img = nib.load(label_file)

        data = np.asarray(img.dataobj)

        data = np.rint(data).astype(np.uint8)

        # 2 -> 1
        data[data == 2] = 1

        new_img = nib.Nifti1Image(
            data,
            img.affine,
            img.header.copy()
        )

        new_img.set_data_dtype(np.uint8)

        nib.save(new_img, label_file)

    print(f"已轉換 {len(needs_convert)} 個 labelsTr 案件：2 -> 1")
```

```python
import os
import time
import subprocess

# ============================================================
# 指定實體 GPU ，VRAM 不足時最多等待 5 分鐘
# ============================================================
GPU = 3  #改成你想要使用的 GPU 編號，例如 0、1、2、3 等
MIN_FREE_GB = 10

def get_vram(gpu):
    result = subprocess.run(
        [
            "nvidia-smi",
            "-i", str(gpu),
            "--query-gpu=memory.free,memory.total",
            "--format=csv,noheader,nounits",
        ],
        capture_output=True,
        text=True,
        check=True,
    )

    free_mb, total_mb = map(
        float, result.stdout.strip().split(",")
    )

    return free_mb / 1024, total_mb / 1024

free_gb, total_gb = get_vram(GPU)

print(f"指定實體 GPU：{GPU}")
print(f"VRAM：{free_gb:.1f} / {total_gb:.1f} GB free")

deadline = time.time() + 300

while free_gb < MIN_FREE_GB and time.time() < deadline:
    print(
        f"剩餘 VRAM 不足 {MIN_FREE_GB} GB，"
        "30 秒後重試..."
    )
    time.sleep(30)
    free_gb, total_gb = get_vram(GPU)
    print(f"目前 VRAM：{free_gb:.1f} / {total_gb:.1f} GB free")

if free_gb < MIN_FREE_GB:
    print(
        f"WARNING：等待 5 分鐘後仍只有 {free_gb:.1f} GB 空餘。"
        "請留意訓練時可能發生 OOM。"
    )
else:
    print(f"實體 GPU {GPU} 的 VRAM 足夠，可以開始訓練。")
```

```python
from pathlib import Path
import shutil

HOME = Path.home()

TEST_SOURCE = (
    HOME / "Public/train/med-ddpm-2/image"
)

IMAGES_TS = (
    HOME
    / "practice_seg/"
      "nnUNet_raw/Dataset101_MedDDPM/imagesTs"
)

existing = sorted(IMAGES_TS.glob("*.nii.gz"))

if len(existing) > 0:
    print(f"imagesTs 已有資料（{len(existing)} cases，全班共享資料集或先前已建立），跳過複製。")
else:
    assert TEST_SOURCE.exists(), (
        f"請確認測試資料來源:\n{TEST_SOURCE}"
    )

    IMAGES_TS.mkdir(
        parents=True,
        exist_ok=True
    )

    test_files = sorted(TEST_SOURCE.glob("*.nii.gz"))

    print("Test files:", len(test_files))

    for idx, src in enumerate(test_files):

        dst = (
            IMAGES_TS
            / f"test_{idx:04d}_0000.nii.gz"
        )

        shutil.copy2(src, dst)

    print("測試資料準備完成。")
```

```python
import os
import hashlib
import sys
import subprocess
from pathlib import Path

#GPU_ID = os.environ.get("COURSE_GPU") or str(
    #int(hashlib.sha256((os.environ.get("USER") or "shared").encode()).hexdigest(), 16) % 8
#)

#os.environ["CUDA_VISIBLE_DEVICES"] = GPU_ID

env = os.environ.copy()
env["CUDA_VISIBLE_DEVICES"] = "3" #改成你想要使用的 GPU 編號，例如 0、1、2、3 等

HOME = Path.home()

NNUNET_RAW = (
    HOME
    / "practice_seg/nnUNet_raw"
)

NNUNET_PREPROCESSED = (
    HOME
    / "practice_seg/nnUNet_preprocessed"
)

NNUNET_RESULTS = (
    HOME
    / "practice_seg/nnUNet_results"
)

os.environ["nnUNet_raw"] = str(NNUNET_RAW)
os.environ["nnUNet_preprocessed"] = str(NNUNET_PREPROCESSED)
os.environ["nnUNet_results"] = str(NNUNET_RESULTS)

INPUT_FOLDER = (
    NNUNET_RAW
    / "Dataset101_MedDDPM"
    / "imagesTs"
)

OUTPUT_FOLDER = (
    HOME
    / "practice_seg"
    / "predictions_3d_fullres"
)

OUTPUT_FOLDER.mkdir(
    parents=True,
    exist_ok=True
)

assert INPUT_FOLDER.exists()

input_files = sorted(
    INPUT_FOLDER.glob("*_0000.nii.gz")
)

assert len(input_files) > 0, (
    "imagesTs 沒有 *_0000.nii.gz"
)

exe = (
    Path(sys.prefix)
    / "bin"
    / "nnUNetv2_predict"
)

cmd = [
    str(exe),

    "-d",
    "101",

    "-i",
    str(INPUT_FOLDER),

    "-o",
    str(OUTPUT_FOLDER),

    "-tr",
    "nnUNetTrainer_10epochs",

    "-c",
    "3d_fullres",

    "-f",
    "0",
]

print("Physical GPU:", env["CUDA_VISIBLE_DEVICES"])
print("Input:", INPUT_FOLDER)
print("Output:", OUTPUT_FOLDER)

print("\nCommand:")
print(" ".join(cmd))

subprocess.run(
    cmd,
    env=os.environ.copy(),
    check=True
)

print("\nInference 完成。")
print("結果:", OUTPUT_FOLDER)
```
